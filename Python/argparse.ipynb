# How can we do this? ```argparse```!
- ```argparse``` package
    - Tutorial: https://docs.python.org/3/howto/argparse.html
    - Documentation: https://docs.python.org/3/library/argparse.html
- What does it do?
    - Map flags to arguments
    - Order of arguments can be supplied in any order
    - Can automatically convert arguments to correct type (i.e. int, float, str)
    - Can specify acceptable choices
    - Arguments representing files can automatically be converted to file handles
    - Create optional switches (boolean arguments that don't accept values)
    - Generates help response

````python
%%bash
scripts/betterscript.py --help 
````

# Setting default values

```python
def get_args():
	parser = argparse.ArgumentParser(description="A program to introduce yourself")
	parser.add_argument("-k", "--kids", help="How many kids you have", type=int, default=0)
	return parser.parse_args()
```

# Giving users a limited set of choices

```python
def get_args():
	parser = argparse.ArgumentParser(description="blah blah blah")
	parser.add_argument("--color", choices=["red","blue","green"])
	return parser.parse_args()
```

# Automatically create file-handles
```python
def get_args():
	parser = argparse.ArgumentParser(description="Plot sliding window GC content of DNA sequence")
	parser.add_argument("-s", "--seq", help="sequence file (FASTA format)", \
        required=True, type=argparse.FileType('r'))
    parser.add_argument("-o", "--out", help="Name of output file", \
        required=True, type=argparse.FileType('w'))
	return parser.parse_args()
```
