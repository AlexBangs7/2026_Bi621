# Dictionary operations
* ```d[key]``` – returns value of dictionary key
* ```d[key] = value``` – assigns value to dictionary key
* ```d[key] += value2``` – increments original value of dictionary key by value2
* ```del d[key]``` – deletes key from dictionary

````python
DNAbases = set('ATGCNatcgn')
RNAbases = set('AUGCNaucgn')

def validate_base_seq(seq,RNAflag=False):
    '''This function takes a string. Returns True if string is composed
    of only As, Ts (or Us if RNAflag), Gs, Cs. False otherwise. Case insensitive.'''
    return set(seq)<=(RNAbases if RNAflag else DNAbases)

def gc_content(DNA):
    '''Returns GC content of a DNA sequence as a decimal between 0 and 1.'''
    assert validate_base_seq(DNA), "String contains invalid characters - are you sure you used a DNA sequence?"
    
    DNA = DNA.upper()
    return (DNA.count("G")+DNA.count("C"))/len(DNA)
````

````python
with open("./files/example.fq","r") as fh:
    i = 0
    for line in fh:
        i+=1
        line = line.strip('\n')
        if i%4 == 2:
            print(line)
            flag = validate_base_seq(line)
            print(flag)
            gc = gc_content(line)
            print(gc)
        
````
