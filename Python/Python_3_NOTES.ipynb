# Rewrite `validate_base_seq()` with sets

````python
def validate_base_seq(seq,RNAflag=False):
    '''This function takes a string. Returns True if string is composed
    of only As, Ts (or Us if RNAflag), Gs, Cs. False otherwise. Case insensitive.'''
    seq = seq.upper()
    return len(seq) == seq.count("A") + seq.count("U" if RNAflag else "T") + seq.count("G") + seq.count("C")

assert validate_base_seq("AATAGAT") == True, "Validate base seq does not work on DNA"
assert validate_base_seq("AAUAGAU", True) == True, "Validate base seq does not work on RNA"
print("Passed DNA and RNA tests")
````

# Iterating over a sequence with a loop
    
## In bash

```bash
for THING in $SEQUENCE
do
    commands
done
```
## In Python

```python
for THING in SEQUENCE:
    statements
```

````python
# Iterate over a list

Barcodes = ["AAAA","TTTT","CCCC","GGGG"]
for barcode in Barcodes:
    print(barcode)
````

````python
# Iterate over a tuple

my_tuple = ("ATCG", 3.5, "increased")
for element in my_tuple:
    print(element)
````

````python
# Iterate over a set

my_set = {"ATCG", 3.5, "increased"}
for element in my_set:
    print(element)
````

````python
# Iterate over a list when you need the index

for i, barcode in enumerate(Barcodes):
    print(f'{i}\t{barcode}')
````
