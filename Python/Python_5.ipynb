````python
def comp_DNA(DNA: str) -> str:
    new_seq = ""
    for base in DNA:
        if base == "A":
            new_seq += "T"
        if base == "T":
            new_seq += "A"
        if base == "G":
            new_seq += "C"
        if base == "C":
            new_seq += "G"
    return new_seq

%timeit comp_DNA("AATTGGCC")
````

````python
def comp_DNA(DNA):
    new_seq = ""
    for base in DNA:
        if base == "A":
            new_seq += "T"
        elif base == "T":
            new_seq += "A"
        elif base == "G":
            new_seq += "C"
        elif base == "C":
            new_seq += "G"
    return new_seq

%timeit comp_DNA("AATTGGCC")
````

````python
my_dict = {"A":"T","T":"A","G":"C","C":"G"}

def comp_DNA(DNA):
    new_seq = ""
    for base in DNA:
        new_seq += my_dict[base]
    return new_seq

%timeit comp_DNA("AATTGGCC")
````

````python
tab = str.maketrans("ACTG", "TGAC")

def comp_DNA(DNA):
    return DNA.translate(tab)

%timeit comp_DNA("AATTGGCC")
````

# Loop with final clause
An ```else``` statement can be added after a loop to be executed after the loop condition evaluates to ```false```

````python
i = 1 
while i < 10:
    print(i, end=" ")
    print("this is running")
    i += 2
else:
    print(i,"\nthe loop is finished")
print("we're out of the LOOOOOP")
````

# Interrupting a loop
* ```continue``` – return to the condition of the loop
* ```break``` – breaks out of the loop entirely (similarly, ```return``` inside a loop within a function)

````python
i = -1
while i < 5:
    i += 2
    print(i, end=" ")
    if i == 3:
        continue
    print("this is running")
print("Done")
````

````python
i = -1
while i < 10:
    print(i, end=" ")
    if i == 3:
        break
    print("this is running")
    i += 2
print("Done")
````

# Comprehension
* Creates a set, list, or dictionary from the results of evaluating an expression for each element of another collection
```python
[expression for item in collection]
```

````python
my_list = [1,2,3,4]
my_list = [x+100 for x in my_list]
print(my_list)    
````

````python
my_list = [1,2,3,4]
for i, x in enumerate(my_list):
    my_list[i] = x+100
print(my_list)
````

# Can re-write `validate_base_seq` using a comprehension

````python
def validate_base_seq(seq: str, RNAflag: bool=False) -> bool:
    valid_bases = 'ACGU' if RNAflag else 'ACGT'
    return all([base in valid_bases for base in seq.upper()])
validate_base_seq("ATTGCGA")
````

# Creating your own module

Nice documentation here: https://docs.python.org/3/tutorial/modules.html
* A module is a file that contains python function definitions and other statements
* To create a module, all we need to do is create a ```.py``` file that contains the functions and statements we're interested in
* We can then ```import``` this file just like we would any other module

````python
#add this function to our NEW module!!!
DNA_bases = set('ATGCNatcgn')
RNA_bases = set('AUGCNaucgn')

def validate_base_seq(seq,RNAflag=False):
    '''This function takes a string. Returns True if string is composed
    of only As, Ts (or Us if RNAflag), Gs, Cs. False otherwise. Case insensitive.'''
    return set(seq)<=(RNA_bases if RNAflag else DNA_bases)

assert validate_base_seq("AATAGAT") == True, "Validate base seq does not work on DNA"
assert validate_base_seq("AAUAGAU", True) == True, "Validate base seq does not work on RNA"
assert validate_base_seq("Hi there!") == False, "Validate base seq fails to recognize nonDNA"
assert validate_base_seq("Hi there!", True) == False, "Validate base seq fails to recognize nonDNA"
print("Passed DNA and RNA tests")
````

# Exception handling
```python
try:
    #try-statements
except:
    #except-statements
```
* If the try statements complete without error, except statement is skipped and program continues
* If try statements cause an error, except-statements executed, prevents program from stopping excecution and printing a traceback

````python
file = 'nonexistant_file.txt'

try:
    1/0
    fh = open(file,'r')
    
except FileNotFoundError:
    print(file, "is non-existant")
except ZeroDivisionError:
    print("no division by zero")
    
print(5+3)
````
